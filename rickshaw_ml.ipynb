```python
# Load the saved pipelines and make a direct example prediction
import joblib, pandas as pd
classification_model = joblib.load('models/best_rickshaw_classification_model.pkl')
regression_model = joblib.load('models/best_rickshaw_regression_model.pkl')
pair = pd.DataFrame([{'pickup_distance_km': 0.8, 'time_difference_min': 5, 'route_overlap_percent': 82, 'detour_km': 0.7, 'destination_compatibility': 0.90, 'preference_compatible': 1}])
print('Match:', classification_model.predict(pair)[0])
print('Estimated detour:', round(regression_model.predict(pair[['pickup_distance_km', 'time_difference_min', 'route_overlap_percent', 'destination_compatibility', 'preference_compatible']])[0], 2), 'km')
```
